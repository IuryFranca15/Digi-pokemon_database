```python
import pandas as pd
import matplotlib
```

```python
df = pd.read_csv('digimonlist.csv')
```

```python
#quantos digimons temos?
qtd = df['Number'].count()
print(qtd)
```

```python
#quantos digimon baby temos?
baby = df['Stage'] == 'Baby'
total = baby.sum()
print(total)
```

```python
#quais são os digimon baby?
df_reduzido = df[['Digimon', 'Stage']]
baby = df_reduzido[df_reduzido['Stage'] == 'Baby'][['Digimon', 'Stage']]
print(baby)
```

```python
df.head()
```

```python
hp950 = df[df['Lv 50 HP'] == 950][['Digimon', 'Lv 50 HP']]
print(hp950)
```

```python
#todos os estágios
estagios = df.groupby('Stage')[['Digimon']].count()
estagios.sort_values('Stage')
```

```python
estagios.plot(kind = 'pie',subplots= True, autopct='%1.1f%%')
```

```python
baby50 = df[df['Stage'] == 'Baby'][['Digimon','Lv50 Atk']]
baby50.sort_values('Lv50 Atk', ascending = False)
```

```python
atributos = df.groupby('Attribute')[['Digimon']].count()
atributos.sort_values('Attribute')
```

```python
atributos.plot(kind = 'pie', subplots= True)
```
